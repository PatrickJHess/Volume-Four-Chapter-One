:class: dropdown

**The Logic:**
Instead of copying and pasting functions or tricking Python into reading raw text from a URL, we are using Python's official package installer (`pip`). We are telling `pip` to bypass the public registry and install our custom quantitative toolkit directly from a GitHub repository, treating it exactly like standard libraries such as `pandas` or `numpy`.

**The Workflow:**
1. **Fetch & Build:** The `%pip install git+https://...` command tells your Jupyter environment to clone the repository from GitHub and install the `financial_quant` package directly into your system's `site-packages` directory.
2. **Import:** `import financial_quant as fq` loads the package into your notebook's memory and assigns it the quick alias `fq`.
3. **Routing:** Behind the scenes, a special file called `__init__.py` acts as the package's "front door." It automatically gathers complex tools from deeply nested folders (like our fixed-income models and chart visualizers) and serves them up directly to the surface.
4. **Execute:** You don't have to worry about where the files live. You just type `fq.one_y_axis()` or `fq.calc_ytm()`, and Python immediately knows where to route the request.

**Why do this?**
This is exactly how professional software engineering teams manage and distribute code. It keeps your notebooks **incredibly clean**, ensures everyone is using the exact same version of the math models, and guarantees your code is 100% portable to any cloud environment.

## 🚀 secure_key_setup makes the FRED API available to the notebook
This function is a smart setup wizard that safely loads your FRED API key into the notebook environment.

* Seamless Loading: If you have already saved your key in a previous session, it automatically finds it and loads it into the background.

* Secure Storage: If your key is missing, it launches an interactive UI to help you safely vault your key (using Google Colab Secrets or hidden local files) so you never accidentally share it.

* Live Validation: Before loading, it actively pings the FRED servers to ensure your key is valid and active, instantly catching typos or formatting errors.

* Cross-Environment: It automatically detects if you are running on Google Colab, a local Jupyter server, or a temporary Binder session, and adapts its security protocols accordingly.


````python
fq.secure_key_setup("fred_key")
````

## 🎬 Demonstrating the class `FredReader`

Now that our secure key is loaded, we can use our custom `FredReader` class to fetch data.

Because `FredReader` is designed to be smart and efficient, it saves the data it downloads into a local cache. This means if you run this notebook again tomorrow, it won't waste time (or API rate limits) re-downloading old data!

> **📁 Google Colab Users:** To use the cache, you will be prompted to grant this notebook access to your Google Drive. This allows the script to create a folder to save your downloaded CSVs.

In the cell below, we will:
1. Create `fred_data`, an instance of `FredReader`.
2. Define a list of `series_ids` containing one fake ID (`'CPI'`) and one real ID (`'PPIACO'` - Producer Price Index).
3. Call the `.get_series()` method. Notice how it gracefully skips the fake ID without crashing!

````python
fred_data=fq.FredReader()
series_ids=['CPI','PPIACO']
fred_data.get_series(series_ids,start_date='1960-01-01',end_date='2026-01-31')
````
